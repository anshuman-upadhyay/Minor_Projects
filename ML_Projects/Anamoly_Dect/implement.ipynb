```python
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer(as_frame=True)
df = data.frame
```

```python
x = df.iloc[:,:2]
```

```python
x
```

```python
import matplotlib.pyplot as plt
plt.scatter(df.iloc[:,0],df.iloc[:,1])
```

```python
from sklearn.ensemble import IsolationForest
```

```python
clf = IsolationForest(contamination= "auto")
clf.fit(x)
pred = clf.predict(x)
```

```python
pred
```

```python
import numpy as np
index = np.where(pred == -1)[0]
index
```

```python
#mark the normal points
plt.scatter(
    x.iloc[:,0],
    x.iloc[:,1],
    label = "Normal")

#mark anamolies 
plt.scatter(
    x.iloc[index,0],
    x.iloc[index,1],
    color = "red",
    label = "Normal"
    )
```
