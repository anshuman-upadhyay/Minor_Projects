```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets
```

```python
#get the iris data set
iris = datasets.load_iris()
iris
```

```python
iris_data = pd.DataFrame(iris.data)
iris_data.columns = iris.feature_names
iris_data
```

```python
from sklearn.model_selection import train_test_split as tts
x_train,x_test= tts(iris_data,random_state= 35,test_size= 0.23)
```

```python
from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
```

```python
#scale the train data
x_scaled = scalar.fit_transform(x_train)
x_scaled
```

```python
#scale the test data
x_scaled = scalar.fit_transform(x_test)
x_scaled
```

```python
#to convert 4 features into 2 feature we ause dimnetionality reduction 
from sklearn.decomposition import PCA 
pca = PCA(n_components= 2)
pca_train_scaled = pca.fit_transform(x_train)
pca_test_scaled = pca.transform(x_test)
```

```python
pca_train_scaled.shape
```

```python
plt.scatter(pca_train_scaled[:,0],pca_train_scaled[:,1])
```

```python
import scipy.cluster.hierarchy as sc
#dendogram creation
plt.Figure(figsize= (20,7))
plt.title("Dendogram")
sc.dendrogram(sc.linkage(pca_train_scaled,method= "ward"))
plt.title("Dendogram")
plt.xlabel("Sample Index")
plt.ylabel("Euclidean Distance")
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(
    n_clusters= 2 ,
      metric= "euclidean",
      linkage= "ward")
cluster.fit(pca_train_scaled)
```

```python
cluster.labels_
```

```python
plt.scatter(pca_train_scaled[:,0],pca_train_scaled[:,1],c = cluster.labels_ ) 
```

```python
# get the silhoutte score 
#performace metrics
from sklearn.metrics import silhouette_score
silhouette_coeff = []
for k in range(2,15):
    aggo = AgglomerativeClustering(
    n_clusters= k,
      metric= "euclidean",
      linkage= "ward") # make sure the random points are properly distanced
    aggo.fit(pca_train_scaled)
    score = silhouette_score(pca_train_scaled,aggo.labels_)
    silhouette_coeff.append(score)
silhouette_coeff
```

```python
plt.plot(range(2,15),silhouette_coeff)
plt.xticks(range(2,15))
plt.xlabel("Number of clusters")
plt.ylabel("Wcss")
plt.show
```
