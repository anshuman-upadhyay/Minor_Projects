```python
from sklearn.datasets import load_breast_cancer
import numpy as np
```

```python
#load the data set
data = load_breast_cancer()
data
```

```python
#get the independent and dependent variables out
x,y = data["data"],data["target"]
x
```

```python
# split test and train data set to remove data leak
from sklearn.model_selection import train_test_split as tts
x_train,x_test,y_train,y_test = tts(
    x,y,
    random_state= 43,
    test_size= 0.24
)
x_test
```

```python
#standardize the entire data 
from sklearn.preprocessing import StandardScaler 
scalar = StandardScaler()
x_train = scalar.fit_transform(x_train)
x_test = scalar.transform(x_test)
```

```python
from sklearn.svm import SVC
model = SVC(
    kernel = "sigmoid"
)
model.fit(x_train,y_train)
```

```python
#get accuracy 
from sklearn.metrics import accuracy_score,confusion_matrix,roc_auc_score,classification_report
y_pred = model.predict(x_test)
acc = accuracy_score(y_test,y_pred)
cm = confusion_matrix(y_test,y_pred)
ra = roc_auc_score(y_test,y_pred)
clas = classification_report(y_test,y_pred)

print(f"Accuracy : {acc}")
print(f"Confusion Matrix : {cm}")
print(f"ROC_AUC_Score : {ra}")
print(f"Classification Report : {clas}")

```
