```python
sen = [
    "Machine learning models improve their performance by learning from data over time.",
    
    "I enjoy solving coding problems on weekends to improve my programming skills.",
    
    "Natural language processing helps computers understand and generate human language.",
    
    "The student completed the assignment before the submission deadline.",
    
    "Artificial intelligence is transforming industries such as healthcare and finance.",
    
    "Reading technical books regularly can strengthen problem-solving abilities.",
    
    "The restaurant served delicious food and the staff was very friendly.",
    
    "Deep learning models often require large amounts of training data.",
    
    "My laptop battery drains quickly when I run multiple virtual machines together.",
    
    "The football match was dramatic final goal.",
    
    "Data preprocessing is an important step before training a machine learning model.",
    
    "The programmer fixed the bug after carefully analyzing the error logs.",
    
    "Cloud computing resources over the internet efficiently.",
    
    "I started learning Rust because of its focus on safety and performance.",
    
    "The movie had an interesting storyline and excellent visual effects.",
    
    "Students should practice algorithms regularly to perform well in coding interviews.",
    
    "The train arrived late due to heavy rainfall in the city.",
    
    "Cybersecurity is becoming increasingly important in the modern digital world.",
    
    "The new smartphone camera even in low light."
]
```

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sen #Each sentances have different lengths
```

```python
#Set the vocabulary size  : HYPERPARAMETER
voc_size = 10000 #Sets all the encoded values between 0 - 10000
```

```python
# vec = tf.keras.layers.TextVectorization(
#     max_token = voc_size,
#     output_mode = "int",
#     output_sequence_length = 5 
# )
```

```python
one_hot_rep =[ one_hot(words,voc_size ) for words in sen]
print(*one_hot_rep,end= "\n")
```

Perform Padding :

Make all the lengths of the input sentences equal by adding 0s in front or back of the vector arrays

```python
import numpy as np
from tensorflow.keras.layers import Embedding
from keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
sent_length = 15 # Set max sentence lenght 
embedded_docs =pad_sequences(one_hot_rep,padding = "pre",maxlen = sent_length)

print(embedded_docs)
```

Perform Feature REpresentation

```python
dim = 10# this is the number of dimensions we use
```

```python
model = Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))

model.compile("adam","mse") #adam optimizer,mse loss
```

```python
model.summary()
```

```python
temp = []
for i in range(len(embedded_docs)):
    temp.append(model.predict(embedded_docs[i]))
```
