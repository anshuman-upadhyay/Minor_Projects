```python
#imports
import pandas as pd
```

```python
df = pd.read_csv("train.csv", encoding="latin1",engine = "python")
df
```

```python
df.isnull().sum()
```

```python
#use the contents as the training data
message = df["article_title"]
message 
```

TexT Preprocessing 1

```python
import nltk
#regularize the text
import re
#remove stopwords
from nltk.corpus import stopwords
nltk.download("stopwords")
#lemmatize
from nltk.stem import WordNetLemmatizer
```

```python
#clean out text and use them 
stop_words = stopwords.words("english")
lemmatizer = WordNetLemmatizer()
corpus = []
for i in range(len(message)):
    tokens = re.sub("[^a-zA-Z0-9]", " ", message[i])
    tokens = tokens.lower()
    tokens = tokens.split()

    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
        if word not in stop_words
    ]

    corpus.append(" ".join(tokens))
```

```python
corpus
```

```python
#Prepare the depenent and independent data 
X_text = corpus
y = pd.get_dummies(df["labels"]).iloc[:,1].values
#Properly encode the data
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
y
```

```python
#split the testing and training data here to prevent leakabge
from sklearn.model_selection import train_test_split as tts

x_train,x_test,y_train,y_test = tts(
    X_text,
    y,
    test_size= 0.23,
    random_state=23,
    stratify= y)
```

```python
has = {}
for i in y_train:
    has[i] = has.get(i,0) +1
has 
```

Text Preprocessing 2 

```python
#Preform vectorization of training and testing dependent data individually
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils  import pad_sequences

#set the vocabulary size 
voc_size = 20000
#Set the tokenizer
tokenizer = Tokenizer(num_words = voc_size)
tokenizer.fit_on_texts(x_train)

# #Now tokenize thhe train and test
x_train_seq = tokenizer.texts_to_sequences(x_train )
x_test_seq = tokenizer.texts_to_sequences(x_test )

#apply padding
max_len = 500

x_train_pad = pad_sequences(
    x_train_seq,
    maxlen = max_len,
    padding = "post"

)
x_test_pad = pad_sequences(
    x_test_seq,
    maxlen = max_len,
    padding = "post"

)

x_train_pad
```

Building the BiLSTM"

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Embedding,
    Bidirectional,
    LSTM,
    Dense,
    Dropout
)
```

Pipeling

```python
models = Sequential(
    [ #Embedding
        Embedding(
            input_dim=voc_size,
            output_dim= 128,
            mask_zero= True,

        ),
        #Apply model
        Bidirectional(
            LSTM(
                128, #number of memory cells
                dropout = 0.2,
                recurrent_dropout = 0.2
            )

        ),
        Dropout(0.3),#Randomly deacivate 30% neurons
        #hidden layes with relu
        Dense(128,activation="relu"),
        #output layer
        Dense(1,activation="sigmoid")
    ]
)
```

```python
models.compile(
    loss = "binary_crossentropy",
    optimizer = "adam",
    metrics = ["accuracy"]

)
```

```python
history = models.fit(
    x_train_pad,
    y_train,
    validation_data = (x_test_pad, y_test),
    epochs = 15,
    batch_size =45
)
```

```python
#Accuracy testing
loss, acc= models.evaluate(
    x_test_pad,
    y_test
)
print("Accuracy : ",acc)
```
